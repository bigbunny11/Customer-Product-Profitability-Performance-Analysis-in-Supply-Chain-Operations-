Revenue & Profit Overview (KPIs)

```python
total_revenue = df['Sales'].sum()
total_profit  = df['Order Profit Per Order'].sum()
profit_margin = (total_profit / total_revenue) * 100
avg_discount  = df['Order Item Discount Rate'].mean() * 100

print(f"Total Revenue:   ${total_revenue:,.2f}")
print(f"Total Profit:    ${total_profit:,.2f}")
print(f"Profit Margin:   {profit_margin:.2f}%")
print(f"Avg Discount:    {avg_discount:.2f}%")
```

Customer Analysis

```python
customer_df = df.groupby('Customer Id').agg(
    Total_Sales   = ('Sales', 'sum'),
    Total_Profit  = ('Order Profit Per Order', 'sum'),
    Order_Count   = ('Sales', 'count')
).reset_index()

customer_df['Profit_Margin'] = (customer_df['Total_Profit'] / customer_df['Total_Sales']) * 100

# Top 10 most profitable customers
print(customer_df.sort_values('Total_Profit', ascending=False).head(10))

# Loss-making customers
print("Loss-making customers:", (customer_df['Total_Profit'] < 0).sum())
```

Product & Category Profitability

```python
category_df = df.groupby('Category Name').agg(
    Revenue = ('Sales', 'sum'),
    Profit  = ('Order Profit Per Order', 'sum')
).reset_index()
category_df['Margin_%'] = (category_df['Profit'] / category_df['Revenue']) * 100
print(category_df.sort_values('Margin_%', ascending=False))
```

Discount Impact

```python
# Bin discount rates and see effect on profit ratio
df['Discount_Bin'] = pd.cut(df['Order Item Discount Rate'],
                             bins=[0, 0.05, 0.1, 0.2, 0.3, 1.0],
                             labels=['0-5%','5-10%','10-20%','20-30%','30%+'])

df.groupby('Discount_Bin')['Order Item Profit Ratio'].mean().plot(kind='bar')
```

Regional / Market Analysis

```python
market_df = df.groupby('Market').agg(
    Revenue = ('Sales', 'sum'),
    Profit  = ('Order Profit Per Order', 'sum')
).reset_index()
market_df['Margin_%'] = (market_df['Profit'] / market_df['Revenue']) * 100
print(market_df.sort_values('Margin_%', ascending=False))
```
